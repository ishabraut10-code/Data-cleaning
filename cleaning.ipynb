```python
import pandas as pd

df=pd.read_csv("netflix_titles.csv")

print("Sample Head")
print(df.head())


print("Data set info ")
print(df.info())

print("Summary Statistics")
print(df.describe(include="all"))

```

```python
df.drop_duplicates()
```

```python
df.isnull().sum()



missing_values = df.isnull().sum().sort_values(ascending=False)
missing_values

missing_percent = (df.isnull().sum() / len(df)) * 100
missing_percent.sort_values(ascending=False)

```

```python
df['director']=df['director'].fillna("Unkown")
df['cast']=df['cast'].fillna("Unkown")
df['country']=df["country"].fillna("Unkown")

df["date_added"] = df["date_added"].fillna(df["date_added"].mode()[0])
df['rating']=df['rating'].fillna(df["rating"].mode()[0])

df=df.dropna(subset=("duration"))
```

```python
df[["duration_value", "duration_type"]] = df["duration"].str.split(" ", expand=True)
df["duration_value"] = pd.to_numeric(df["duration_value"], errors="coerce")
df["duration_type"] = df["duration_type"].replace({
    "min": "Minutes",
    "Season": "Seasons",
    "Seasons": "Seasons"
})
```

```python

df["date_added"] = pd.to_datetime(df["date_added"], errors="coerce")
```

```python
df.to_csv('cleaned-data.csv')
```
